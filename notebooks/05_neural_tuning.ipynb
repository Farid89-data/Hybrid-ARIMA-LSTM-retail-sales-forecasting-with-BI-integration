Optuna trials (tree-structured Parzen estimator, 15 trials) and the selected configurations.

```python
import sys
from pathlib import Path
CODE_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(CODE_DIR))
import json
import numpy as np
import pandas as pd
from IPython.display import Image, Markdown, display
from src import config, experiment
pd.set_option("display.width", 160)
pd.set_option("display.max_columns", 30)
TABLES = config.OUTPUT_DIR / "tables"
FIGS = config.FIGURE_DIR
```

```python
rows = []
for name in ['m5', 'rossmann']:
    for kind in experiment.NEURAL_KINDS:
        p = config.METRIC_DIR / f'{name}_{kind}_best_params.json'
        if p.exists():
            d = json.loads(p.read_text())
            rows.append({'dataset': name, 'model': kind, **d['best_params'], 'validation_mase': d['best_value']})
pd.DataFrame(rows)
```

```python
for name in ['m5', 'rossmann']:
    for kind in ['ffnn', 'lstm', 'hyb_d', 'hyb_ms']:
        p = config.METRIC_DIR / f'{name}_{kind}_tuning_trials.csv'
        if p.exists():
            t = pd.read_csv(p)
            print(name, kind, 'trials:', len(t), 'best:', round(t['value'].min(), 4), 'median:', round(t['value'].median(), 4))
```
